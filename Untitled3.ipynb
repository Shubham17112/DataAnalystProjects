```python
# =======================================
# Imports & Setup
# =======================================
from google.colab import drive
drive.mount('/content/drive')

BARK_CACHE_PATH = "/content/drive/MyDrive/bark_models"
import os
os.makedirs(BARK_CACHE_PATH, exist_ok=True)
os.environ["HF_HOME"] = BARK_CACHE_PATH
os.environ["TRANSFORMERS_CACHE"] = BARK_CACHE_PATH
os.environ["BARK_CACHE_DIR"] = BARK_CACHE_PATH

import importlib.util
import random, numpy as np, torch, re
import soundfile as sf
from bark import SAMPLE_RATE, generate_audio, preload_models
from pydub import AudioSegment
from pydub.effects import normalize
from IPython.display import Audio, display

# =======================================
# Install dependencies if missing
# =======================================
def ensure_installed():
    required = ["bark", "torch", "torchaudio", "torchvision", "indic_transliteration"]
    for pkg in required:
        if importlib.util.find_spec(pkg) is None:
            print(f"📦 Installing {pkg}...")
            if pkg == "bark":
                !pip install git+https://github.com/suno-ai/bark.git soundfile pydub indic-transliteration
            else:
                !pip install {pkg}

ensure_installed()

# =======================================
# Seed system
# =======================================
VOICE_SEEDS = {
    "female_soft": 42,
    "female_clear": 123,
    "female_warm": 456,
    "male_deep": 789,
    "male_friendly": 101,
    "narrator_pro": 202,
    "storyteller": 303,
    "horror_voice": 666,
}

def set_seed(seed_value=303):
    random.seed(seed_value)
    np.random.seed(seed_value)
    torch.manual_seed(seed_value)
    torch.cuda.manual_seed(seed_value)
    torch.cuda.manual_seed_all(seed_value)
    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False
    os.environ['PYTHONHASHSEED'] = str(seed_value)

# =======================================
# Preload Bark Models
# =======================================
if not (os.path.exists(BARK_CACHE_PATH) and any(f.endswith(".pt") for f in os.listdir(BARK_CACHE_PATH))):
    print("⬇️ Downloading Bark models...")
    preload_models()
else:
    print("✅ Bark models already cached.")

# =======================================
# Utility: Split text into chunks
# =======================================
def split_text(text, max_chars=150):
    sentences = re.split('(?<=[।.!?]) +', text.strip())
    chunks, current = [], ""
    for sentence in sentences:
        if len(current) + len(sentence) <= max_chars:
            current += " " + sentence if current else sentence
        else:
            if current: chunks.append(current.strip())
            current = sentence
    if current: chunks.append(current.strip())
    return [c for c in chunks if c.strip()]

# =======================================
# Core: Generate audio
# =======================================
def generate_with_seed(text, seed_name="female_soft", custom_seed=None, history_prompt=None):
    seed_value = custom_seed or VOICE_SEEDS.get(seed_name, 42)
    set_seed(seed_value)
    return generate_audio(text, history_prompt=history_prompt, text_temp=0.6, waveform_temp=0.6)

# =======================================
# Generate multi-voice audiobook with history_prompt
# =======================================
def generate_multi_voice_audiobook(text, voice_mapping=None, history_mapping=None, play_directly=True):
    if voice_mapping is None:
        voice_mapping = {"[Horror]": "horror_voice", "[whispering]": "female_soft", "default": "narrator_pro"}
    if history_mapping is None:
        history_mapping = {"default": None}

    chunks = split_text(text)
    final_audio = AudioSegment.silent(duration=0)

    for i, chunk in enumerate(chunks, start=1):
        chosen_voice = voice_mapping.get("default")
        chosen_history = history_mapping.get("default")
        for keyword, voice in voice_mapping.items():
            if keyword in chunk and keyword != "default":
                chosen_voice = voice
                chosen_history = history_mapping.get(keyword, None)
                break

        print(f"🎙 Chunk {i}: Using voice '{chosen_voice}' and history '{chosen_history}' for: {chunk[:30]}...")
        audio_array = generate_with_seed(chunk, seed_name=chosen_voice, history_prompt=chosen_history)

        temp_wav = f"chunk_{i}.wav"
        sf.write(temp_wav, audio_array, SAMPLE_RATE)
        chunk_audio = normalize(AudioSegment.from_wav(temp_wav)) - 3
        if i < len(chunks):
            chunk_audio += AudioSegment.silent(duration=500)
        final_audio += chunk_audio
        os.remove(temp_wav)

    if play_directly:
        temp_file = "temp_audio.wav"
        final_audio.export(temp_file, format="wav")
        display(Audio(temp_file, rate=SAMPLE_RATE))
    else:
        return final_audio

# =======================================
# Example usage
# =======================================
story_text = """
[fav] Hello friends! Today we are going to hear a new story!
This house is very old.

"""

generate_multi_voice_audiobook(
    story_text,
    # voice_mapping={"[Horror]": "horror_voice", "[whispering]": "female_soft", "default": "male_deep"},
    history_mapping={"[fav]":"v2/en_speaker_6",  "default": "v2/en_speaker_5"}
)
```
